```python
print("""This is David's fruit shop.
1: Apple (price : 5000 won)
2: Grape (price : 6000 won)
3: Melon (price : 8000 won)
4: Orange (price: 2000 won)"""
     )

total_price=0
fruit=int(input("enter num between 1-4"))
quantity=int(input("enter the num between 1-10") )

if fruit==1:
    fruit="Apple"
    price=5000
elif fruit==2:
    fruit="Grape"
    price=6000
elif fruit==3:
    fruit="Melon"
    price=8000
else:
    fruit="Orange"
    price=2000

print("""fruit selected""",fruit,
      """price""",price,
     """Quantity""",quantity)
total_price=price*quantity
print("total price is",total_price)

money=int(input("please input the money"))
if money<total_price:
    print("not enough money")
else:
    change=money-total_price
    print(money ,"recived your change is",change)


```
