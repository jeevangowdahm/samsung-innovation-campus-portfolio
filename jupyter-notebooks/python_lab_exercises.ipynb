```python
#2a
num = int(input("Enter the Fibonacci sequence length to be generated : ")) 
firstTerm = 0
secondTerm = 1
print("The Fibonacci series with", num, "terms is :") 
print(firstTerm, secondTerm, end=" ") 
for i in range(2,num):
  curTerm = firstTerm + secondTerm 
  print(curTerm, end=" ")
  firstTerm = secondTerm 
  secondTerm = curTerm
```

```python
#3a
from math import sqrt 

myList = []
num = int(input("Enter the number of elements: ")) 

# 1. Indent the append so every number is saved
for i in range(num):
    val = int(input("Enter the element: ")) 
    myList.append(val)

print('List Contents', myList)

# 2. Calculate Mean after the loop finishes
total = 0
for elem in myList:
    total += elem 
mean = total / num

# 3. Calculate Variance after the loop finishes
total = 0
for elem in myList:
    total += (elem - mean) ** 2 
variance = total / num

stdDev = sqrt(variance) 

print("Mean =", mean) 
print("Variance =", variance)
print("Standard Deviation =", "%.2f" % stdDev)
```

```python
#5a
# Program to sort 6 subject marks in descending order using Bubble Sort
marks = []
print("Enter marks for 6 subjects:")

for i in range(6):
    m = int(input(f"Enter mark {i+1}: "))
    marks.append(m)

print("\nOriginal Marks:", marks)

# Bubble Sort logic
n = len(marks)
for i in range(n):
    for j in range(0, n - i - 1):
        if marks[j] < marks[j+1]:  # Swap if the current mark is smaller than the next
            marks[j], marks[j+1] = marks[j+1], marks[j]

print("Marks from Highest to Lowest:", marks)
```

```python
#6a
# Program to sort the contents of a text file and write to another file

try:
    # 1. Open the source file for reading
    with open("input.txt", "r") as infile:
        # Read lines, strip whitespace, and skip empty lines
        lines = [line.strip() for line in infile if line.strip()]

    # 2. Sort the lines alphabetically
    # Use key=str.lower to ensure a proper A-Z sort regardless of capitalization
    lines.sort(key=str.lower)

    # 3. Open the destination file for writing
    with open("output.txt", "w") as outfile:
        for line in lines:
            outfile.write(line + "\n")

    print("Sorting complete. Check 'output.txt' for the results.")

except FileNotFoundError:
    print("Error: 'input.txt' not found. Please create the file and try again.")
except Exception as e:
    print(f"An unexpected error occurred: {e}")
```
