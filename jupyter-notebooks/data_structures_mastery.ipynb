```python
items = {"Coffee": 7, "Pen": 3, "Paper cup": 2, "Milk": 1, "Coke": 4, "Book": 5}

while True:
    choice = input("Select menu 1)check stock 2)warehousing 3) release 4) exit :")
    
    if choice == "1":
        
        name = input("[check stock] Enter item: ").strip().lower()
        
        
        for key in items:
            if key.lower() == name:
                print(f"Stock: {items[key]}")
                break
                
    elif choice == "2":
       
        user_input = input("[Warehousing] Enter item and quantity: ").strip().split()
        item_name = user_input[0].lower()
        qty = int(user_input[1])
        
        for key in items:
            if key.lower() == item_name:
                items[key] += qty
                break
                
    elif choice == "3":
        
        user_input = input("[Release] Enter item and quantity: ").strip().split()
        item_name = user_input[0].lower()
        qty = int(user_input[1])
        
        for key in items:
            if key.lower() == item_name:
                items[key] -= qty
                break
                
    elif choice == "4":
        print("Program exited.")
        break
```

```python
# Initial inventory dictionary
items = {"Coffee": 7, "Pen": 3, "Paper cup": 2, "Milk": 1, "Coke": 4, "Book": 5}

# Create a mapping of lowercase names to the exact dictionary keys for case-insensitive lookup
key_mapping = {k.lower(): k for k in items.keys()}

while True:
    # Display menu and get user choice
    choice = input("Select menu 1)check stock 2)warehousing 3) release 4) exit :").strip()
    
    if choice == "1":
        # Check stock
        item_input = input("[check stock] Enter item: ").strip().lower()
        if item_input in key_mapping:
            actual_key = key_mapping[item_input]
            print(f"Stock: {items[actual_key]}")
        else:
            print("Item not found in inventory.")
            
    elif choice == "2":
        # Warehousing (increase stock)
        user_input = input("[Warehousing] Enter item and quantity: ").strip().split()
        if len(user_input) >= 2:
            # Handle item names with spaces if needed, but assuming item and count split by last element
            quantity_str = user_input[-1]
            item_name = " ".join(user_input[:-1]).lower()
            
            if quantity_str.isdigit():
                quantity = int(quantity_str)
                if item_name in key_mapping:
                    actual_key = key_mapping[item_name]
                    items[actual_key] += quantity
                    print(f"Warehoused successfully. New stock of {actual_key}: {items[actual_key]}")
                else:
                    # Optional: Add new item to inventory if it doesn't exist
                    new_key = item_name.capitalize()
                    items[new_key] = quantity
                    key_mapping[item_name] = new_key
                    print(f"Added new item {new_key} with stock: {quantity}")
            else:
                print("Invalid quantity entered.")
        else:
            print("Invalid input format. Use: item_name quantity")
            
    elif choice == "3":
        # Release (decrease stock)
        user_input = input("[Release] Enter item and quantity: ").strip().split()
        if len(user_input) >= 2:
            quantity_str = user_input[-1]
            item_name = " ".join(user_input[:-1]).lower()
            
            if quantity_str.isdigit():
                quantity = int(quantity_str)
                if item_name in key_mapping:
                    actual_key = key_mapping[item_name]
                    if items[actual_key] >= quantity:
                        items[actual_key] -= quantity
                    else:
                        print(f"Not enough stock. Current stock is {items[actual_key]}.")
                else:
                    print("Item not found in inventory.")
            else:
                print("Invalid quantity entered.")
        else:
            print("Invalid input format. Use: item_name quantity")
            
    elif choice == "4":
        # Exit the program
        print("Program exited.")
        break
        
    else:
        print("Invalid choice. Please select 1, 2, 3, or 4.")
```
