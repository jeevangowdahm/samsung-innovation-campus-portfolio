```python
import pandas as pd
import numpy as np

data = {
    'day': ['mon', 'tue', 'wed', 'thu', 'fri'],
    'temp': [22, np.nan, 25, np.nan, 20]
}
df = pd.DataFrame(data)

# Apply ffill first, then bfill to catch both NaN values correctly
df['temp_filled'] =df['temp'].ffill().bfill()

print(df)
```

```python
import pandas as pd
import numpy as np

data = {
    'day': ['mon', 'tue', 'wed', 'thu', 'fri'],
    'temp': [22, np.nan, 25, np.nan, 20]
}
df = pd.DataFrame(data)

# Create a copy for filled values
df['temp_filled'] = df['temp']

# Forward fill Tuesday using Monday's value
df['temp_filled'] = df['temp_filled'].ffill()

# Backward fill Thursday using Friday's value
df['temp_filled'] = df['temp_filled'].bfill()

print(df)
```

```python
!pip install yfinance 
import pandas as pd
import datetime
from datetime import date, datetime, time
import matplotlib.pyplot as plt

import yfinance as yf
start = datetime(2025, 1, 1)
end = datetime(2025, 11,30)

sec = yf.download(tickers=['005930.KS'],
start=start,
end=end,
auto_adjust=False)

sec.head()
sec.info()
ma5 = sec['Adj Close'].rolling(window=5).mean()
ma5.head(10)
```

```python
close = sec['Adj Close']

ma5 = sec['Adj Close'].rolling(window=5).mean()
ma10 = sec['Adj Close'].rolling(window=10).mean()
ma60 = sec['Adj Close'].rolling(window=60).mean()
ma120 = sec['Adj Close'].rolling(window=120).mean()

plt.figure(figsize = (15,10))

plt.plot(close, label='close', linewidth=4)
plt.plot(ma5, label='5 window')
plt.plot(ma10, label='10 window')
plt.plot(ma60, label='60 window')
plt.plot(ma120, label='120 window')

plt.legend ()
plt. title('Rolling window calculations')
plt.xlabel('Date')
plt.ylabel('close price')
```
