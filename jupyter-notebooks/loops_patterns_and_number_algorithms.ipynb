```python
bread=["r b","w","wh"]
vegies=["l","t","c"]
saucse=["m","t s","o l"]

athlete_protein={
    "strength a":["beef","chick","tofu"],
    "endurance a":["chick","tofu"],
    "agility a":["chick","mushroom","tofu"]}
print("sandwich combinations are")
for athlete,protein in athlete_protein.items():
    print(f"[{athlete}] options:")
    for b in bread:
        for p in protein:
            for v in vegies:
                for s in saucse:
                    print(f"{b}+{p}+{v}+{s}")
```

```python

current_pos = 0 
days = 1
move_up = 30
move_down = 20
destination =30
current_pos += move_up 

print('day :', days, "Snail's position ", current_pos, 'meters')
while(current_pos <= destination):
    current_pos-=move_down
    days+=1
    current_pos+=move_up
    print('day :', days, "snail's position :", current_pos, 'meters')
print()
print(f'It took {days} days to get out of the well.')
```

```python
person1 = ['David Doe', 20, 1, 180.0, 100.0]
person2 = ['John Smith', 25, 1, 170.0, 70.0]
person3 = ['Jane Carter', 22, 0, 169.0, 60.0]
person4 = ['Peter Kelly', 40, 1, 150.0, 50.0]

person_list = person1 + person2 + person3 + person4

n_persons = int (len(person_list) / 5)
age_sum = 0.0
for age in person_list[1 :: 5]:
    age_sum += age
average_age = float(age_sum) / n_persons
print('the average age is', str(average_age))
print(person_list)
```

```python

for num in range(2, 13):
    is_prime = True
    
    
    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break 
            

    if is_prime:
        print(f"{num} is prime")
    else:
        print(f"{num} is composite")
```
