```python
pip install groq
pip install gradio

from groq import Groq
API="your_groq_api_key_here"
client = Groq(api_key=API)
completion = client.chat.completions.create(
    model="openai/gpt-oss-120b",
    messages=[
      {
        "role": "user",
        "content": "what is ai and respond me in english"
      }
    ],
    temperature=2,
    max_completion_tokens=9696,
    top_p=1,
    reasoning_effort="medium",
    stream=True,
    stop=None
)

for chunk in completion:
    print(chunk.choices[0].delta.content or "", end="")


def chat(message, history):

    messages = [
        {
            "role": "system",
            "content": "You are a AI.respond in kannada script"
        }
    ]

    # Keep only fields supported by Groq
    for msg in history:
        messages.append({
            "role": msg["role"],
            "content": msg["content"]
        })

    messages.append({
        "role": "user",
        "content": message
    })

    response = client.chat.completions.create(
        model="openai/gpt-oss-120b",
        messages=messages
    )

    return response.choices[0].message.content


import gradio as gr
demo = gr.ChatInterface(
    fn=chat,
    title="JJGHM's Chatbot"
)

demo.launch()
```
