```python
def caesar_cipher_encrypt(text):
    result = ""
    for char in text:
        if char.isupper():
            # Get the ASCII value of the character
            char_code = ord(char)
            # Shift the character by 3, wrapping around if necessary
            shifted_code = ((char_code - ord('A') + 3) % 26) + ord('A')
            result += chr(shifted_code)
        else:
            result += char
    return result

# Example usage:
input_sentence = "HELLO WORLD"
encrypted_sentence = caesar_cipher_encrypt(input_sentence)
print(encrypted_sentence)  # Output: KHOOR ZRUOG
```

```python
import random
questions = [
   {
       "question": "What is the capital of France?",
       "choices": ["London", "Berlin", "Paris", "Madrid"],
       "correct": "Paris"
   },
   {
       "question": "Who painted the Mona Lisa?",
       "choices": ["Vincent van Gogh", "Leonardo da Vinci", "Pablo Picasso", "Claude Monet"],
       "correct": "Leonardo da Vinci"
   },
   {
       "question": "Which keyword is used to create a function in Python?",
       "choices": ["func", "def", "function", "lambda"],
       "correct": "def"
   }
]
def run_quiz(questions):
   score = 0
   total = len(questions)
   random.shuffle(questions)
   for idx, q in enumerate(questions, start=1):
       print(f"\nQuestion {idx}: {q['question']}")
       choices = q["choices"][:]
       random.shuffle(choices)
       for i, choice in enumerate(choices, 1):
           print(f"{i}. {choice}")
       
       while True:
           answer = input("Your answer (number): ")
           if answer.isdigit() and 1 <= int(answer) <= len(choices):
               break
           print("Invalid choice. Please enter a valid number.")
       
       if choices[int(answer) - 1] == q["correct"]:
           print(" Correct!")
           score += 1
       else:
           print(f" Wrong! Correct answer: {q['correct']}")
   print(f"\n Final Score: {score}/{total}")
if __name__ == "__main__":
   run_quiz(questions)
```

```python
def check(expr):
    opening = "[{("
    closing = "]})"
    stack = []

    for char in expr:
        if char in opening:
            stack.append(char)
        elif char in closing:
            if len(stack) == 0:
                return False
            if opening.index(stack.pop()) != closing.index(char):
                return False
    return len(stack) == 0

s = input("Enter a string of parentheses: ")

if check(s):
    print("Valid")
else:
    print("Invalid")
```

```python
def josephus_sequence(n, k):
    sequence = []
    queue = []

    # Insert people into the queue
    for i in range(1, n + 1):
        queue.append(i)

    j = 1

    # Continue until only one person remains
    while len(queue) > 1:
        item = queue.pop(0)

        if j % k == 0:
            sequence.append(item)   # Person is eliminated
        else:
            queue.append(item)      # Move person to rear

        j += 1

    # Add the last remaining person
    sequence.append(queue.pop(0))

    return sequence

N = int(input("Input the number of people (N): "))
K = int(input("Input the number to be skipped (K): "))

print("Josephus sequence:", josephus_sequence(N, K))

```

```python
def bubble_sort(a):
    n = len(a)

    for i in range(0,n - 1):
        for j in range(0,n - i - 1):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
    return a

a = list(map(int, input("Enter elements: ").split()))
print("Sorted list:", bubble_sort(a))


```

```python
def bubble_sort(arr):
    a = arr.copy()  
    n = len(a)
    for i in range(n - 1):
        for j in range(n - i - 1):
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
    return a


def insertion_sort(arr):
    a = arr.copy()
    n = len(a)
    for i in range(1, n):
        key = a[i]
        j = i - 1
        while j >= 0 and a[j] > key:
            a[j + 1] = a[j]
            j -= 1
        a[j + 1] = key
    return a


def selection_sort(arr):
    a = arr.copy()
    n = len(a)
    for i in range(n - 1):
        min_pos = i
        for j in range(i + 1, n):
            if a[j] < a[min_pos]:
                min_pos = j
        a[i], a[min_pos] = a[min_pos], a[i]
    return a


def main():
   
    user_input = input("Enter numbers separated by spaces: ")
    numbers = list(map(int, user_input.split()))

    while True:
        print("\n SORTING MENU     ")
        
        print("1. Bubble Sort")
        print("2. Insertion Sort")
        print("3. Selection Sort")
        print("4. Exit")

        try:
            choice = int(input("\nEnter choice (1-4): "))
        except ValueError:
            print("Invalid input! Please enter a valid number.")
            continue

        if choice == 1:
            sorted_arr = bubble_sort(numbers)
            print(f"\n Bubble Sorted Result: {sorted_arr}")
        elif choice == 2:
            sorted_arr = insertion_sort(numbers)
            print(f"\n Insertion Sorted Result: {sorted_arr}")
        elif choice == 3:
            sorted_arr = selection_sort(numbers)
            print(f"\n Selection Sorted Result: {sorted_arr}")
        elif choice == 4:
            print("\nExiting program. Goodbye!")
            break
        else:
            print(" Invalid choice! Please select an option between 1 and 4.")


if __name__ == "__main__":
    main()
```

```python
def merge_two_lists(list1, list2):
    merged = []
    i = j = 0
  
    while i < len(list1) and j < len(list2):
        if list1[i] <= list2[j]:
            merged.append(list1[i])
            i += 1
        else:
            merged.append(list2[j])
            j += 1
            
  
    merged.extend(list1[i:])
    merged.extend(list2[j:])
    return merged

def multiway_merge(list_of_nums):
    if not list_of_nums:
        return []

    while len(list_of_nums) > 1:
        next_round = []
        for i in range(0, len(list_of_nums), 2):
            if i + 1 < len(list_of_nums):
                merged_pair = merge_two_lists(list_of_nums[i], list_of_nums[i+1])
                next_round.append(merged_pair)
            else:
                next_round.append(list_of_nums[i])
        list_of_nums = next_round
        
    return list_of_nums[0]
    
N = int(input("Input the number of list: "))
list_of_nums = []
for i in range(N):
    nums = list(map(int, input("Input a list of numbers: ").split()))
    print(nums)
    list_of_nums.append(nums)

sorted_list = multiway_merge(list_of_nums)
print("Merged into :", sorted_list)
```

```python
class TrominoSolver:
    def __init__(self, n):
        """Initialize a 2^n x 2^n board."""
        self.size = 2 ** n
        self.board = [[0] * self.size for _ in range(self.size)]
        self.tromino_count = 0
    
    def solve(self, missing_row, missing_col):
        """Solve the tromino tiling problem."""
        self._solve(0, 0, self.size, missing_row, missing_col)
        return self.board
    
    def _solve(self, top, left, size, missing_row, missing_col):
        """
        Recursively tile a size x size board starting at (top, left)
        with one missing square at (missing_row, missing_col).
        """
        if size == 1:
            return  # Base case: single cell (the missing one)
        
        half = size // 2
        self.tromino_count += 1
        tromino_id = self.tromino_count
        
        # Find which quadrant contains the missing square
        # Quadrants: 0=top-left, 1=top-right, 2=bottom-left, 3=bottom-right
        quad = self._get_quadrant(top, left, half, missing_row, missing_col)
        
        # Place tromino covering center squares of 3 quadrants (excluding the one with missing)
        # Center positions of each quadrant
        centers = [
            (top + half - 1, left + half - 1),      # top-left center
            (top + half - 1, left + half),           # top-right center
            (top + half, left + half - 1),           # bottom-left center
            (top + half, left + half),               # bottom-right center
        ]
        
        # Place tromino on the 3 quadrants that DON'T contain the missing square
        for i in range(4):
            if i != quad:
                r, c = centers[i]
                self.board[r][c] = tromino_id
        
        # Recursively solve each quadrant
        # Top-left quadrant
        if quad == 0:
            self._solve(top, left, half, missing_row, missing_col)
        else:
            self._solve(top, left, half, centers[0][0], centers[0][1])
        
        # Top-right quadrant
        if quad == 1:
            self._solve(top, left + half, half, missing_row, missing_col)
        else:
            self._solve(top, left + half, half, centers[1][0], centers[1][1])
        
        # Bottom-left quadrant
        if quad == 2:
            self._solve(top + half, left, half, missing_row, missing_col)
        else:
            self._solve(top + half, left, half, centers[2][0], centers[2][1])
        
        # Bottom-right quadrant
        if quad == 3:
            self._solve(top + half, left + half, half, missing_row, missing_col)
        else:
            self._solve(top + half, left + half, half, centers[3][0], centers[3][1])
    
    def _get_quadrant(self, top, left, half, row, col):
        """Determine which quadrant contains the given position."""
        if row < top + half:
            if col < left + half:
                return 0  # top-left
            return 1  # top-right
        else:
            if col < left + half:
                return 2  # bottom-left
            return 3  # bottom-right
    
    def print_board(self):
        """Print the board with visual representation."""
        cell_width = len(str(self.tromino_count)) + 1
        
        print("\n" + "=" * (self.size * cell_width + 2))
        for row in self.board:
            line = "|"
            for cell in row:
                if cell == 0:
                    line += " " * cell_width
                else:
                    line += f"{cell}".center(cell_width)
            line += "|"
            print(line)
        print("=" * (self.size * cell_width + 2))
    
    def visualize_tromino(self):
        """Create ASCII art showing L-shaped trominoes."""
        colors = ['░', '▒', '▓', '█', '◦', '•', '○', '●', 
                  '◇', '◆', '□', '■', '△', '▲', '▽', '▼']
        
        print("\n" + "─" * (self.size * 2 + 1))
        for row in self.board:
            line = "│"
            for cell in row:
                if cell == -1:
                    line += " ✕"
                else:
                    symbol = colors[cell % len(colors)]
                    line += f" {symbol}"
            line += " │"
            print(line)
        print("─" * (self.size * 2 + 1))


def main():
    print("=" * 60)
    print("       TROMINO TILING PROBLEM SOLVER")
    print("=" * 60)
    print("\nThe Problem:")
    print("  Tile a 2^n × 2^n board with one missing square")
    print("  using only L-shaped trominoes (3 squares each).")
    print("\nAlgorithm: Divide and Conquer")
    print("  Time Complexity: O(4^n) = O(N) where N = total cells")
    print("  Space Complexity: O(2^n) = O(√N) for recursion depth")
    
    # Example 1: 4x4 board
    print("\n" + "=" * 60)
    print("EXAMPLE 1: 4×4 Board (n=2)")
    print("=" * 60)
    
    solver1 = TrominoSolver(2)  # 2^2 = 4x4 board
    solver1.board[1][1] = -1  # Mark missing square
    solver1.solve(1, 1)
    
    print("\nBoard with numbered trominoes:")
    solver1.print_board()
    
    print("\nVisual representation (each symbol = one tromino):")
    solver1.visualize_tromino()
    
    print(f"\nTotal trominoes placed: {solver1.tromino_count}")
    print(f"Theoretical: (4² - 1) / 3 = {(16 - 1) // 3}")
    
    # Example 2: 8x8 board
    print("\n" + "=" * 60)
    print("EXAMPLE 2: 8×8 Board (n=3)")
    print("=" * 60)
    
    solver2 = TrominoSolver(3)  # 2^3 = 8x8 board
    solver2.solve(0, 0)  # Missing square at top-left corner
    
    print("\nBoard with numbered trominoes:")
    solver2.print_board()
    
    print(f"\nTotal trominoes placed: {solver2.tromino_count}")
    print(f"Theoretical: (8² - 1) / 3 = {(64 - 1) // 3}")
    
    # Interactive demo
    print("\n" + "=" * 60)
    print("INTERACTIVE DEMO")
    print("=" * 60)
    
    while True:
        try:
            n = int(input("\nEnter board size exponent n (2^n × 2^n, n=1-5): "))
            if n < 1 or n > 5:
                print("Please enter a value between 1 and 5.")
                continue
            
            size = 2 ** n
            print(f"\nBoard size: {size}×{size}")
            
            print(f"Enter missing square position (row, col) [0-{size-1}]:")
            row = int(input("  Row: "))
            col = int(input("  Col: "))
            
            if not (0 <= row < size and 0 <= col < size):
                print("Invalid position. Using center.")
                row = col = size // 2
            
            print(f"\nSolving {size}×{size} board with missing square at ({row}, {col})...")
            
            solver = TrominoSolver(n)
            solver.solve(row, col)
            
            print("\nSolution:")
            solver.print_board()
            
            print(f"\n✓ Successfully placed {solver.tromino_count} trominoes")
            
        except ValueError:
            print("Invalid input. Please enter integers.")
        except KeyboardInterrupt:
            print("\n\nExiting...")
            break
        
        cont = input("\nTry another configuration? (y/n): ")
        if cont.lower() != 'y':
            break
    
    print("\n" + "=" * 60)
    print("Thank you for using the Tromino Tiling Solver!")
    print("=" * 60)


if __name__ == "__main__":
    main()
```

```python
import sys

def tromino_fill(board, top, left, size, missing_r, missing_c, tile_id_holder):
    """
    Fills a size x size board (top-left corner at top,left) with L-trominoes,
    leaving exactly the cell (missing_r, missing_c) uncovered.
    Each placed tromino is written into `board` using a unique tile number.
    """
    if size == 1:
        return  # single cell is the hole itself, nothing to place

    half = size // 2
    tile_id_holder[0] += 1
    tile_id = tile_id_holder[0]

    # center coordinates
    center_r, center_c = top + half, left + half

    # quadrant boundaries: (row_start, col_start, is_missing_here)
    quadrants = [
        (top, left),                # top-left
        (top, left + half),         # top-right
        (top + half, left),         # bottom-left
        (top + half, left + half),  # bottom-right
    ]

    # figure out which quadrant contains the real missing cell
    def in_quadrant(qr, qc):
        return qr <= missing_r < qr + half and qc <= missing_c < qc + half

    # the four cells adjacent to the exact center, one per quadrant
    center_cells = [
        (center_r - 1, center_c - 1),  # belongs to top-left quadrant
        (center_r - 1, center_c),      # top-right
        (center_r, center_c - 1),      # bottom-left
        (center_r, center_c),          # bottom-right
    ]

    new_missing = [None, None, None, None]

    for i, (qr, qc) in enumerate(quadrants):
        if in_quadrant(qr, qc):
            # this quadrant keeps the real missing cell; don't touch its center cell
            new_missing[i] = (missing_r, missing_c)
        else:
            # place a piece of the tromino at this quadrant's center cell
            r, c = center_cells[i]
            board[r][c] = tile_id
            new_missing[i] = (r, c)  # this becomes the "fake hole" for recursion

    # recurse into all four quadrants
    for i, (qr, qc) in enumerate(quadrants):
        mr, mc = new_missing[i]
        tromino_fill(board, qr, qc, half, mr, mc, tile_id_holder)


def print_board(board):
    n = len(board)
    width = len(str(n * n // 3 + 1)) + 1
    for row in board:
        print("".join(f"{('X' if v == -1 else v):>{width}}" for v in row))


def solve(n_power, missing_r, missing_c):
    """
    n_power: board is 2^n_power x 2^n_power
    """
    size = 2 ** n_power
    if not (0 <= missing_r < size and 0 <= missing_c < size):
        raise ValueError("Missing cell out of bounds")

    board = [[0] * size for _ in range(size)]
    board[missing_r][missing_c] = -1  # -1 marks the hole, stays uncovered

    tile_id_holder = [0]  # mutable counter for unique tromino IDs
    tromino_fill(board, 0, 0, size, missing_r, missing_c, tile_id_holder)

    return board


if __name__ == "__main__":
    # Example: 8x8 board (n_power=3), missing cell at (3, 5)
    n_power = 3
    missing_r, missing_c = 3, 5

    board = solve(n_power, missing_r, missing_c)
    print_board(board)
```

```python
def find_minimum(row, col, triangle):
    if row == len(triangle):
        return 0

    minimum = min(
        find_minimum(row + 1, col, triangle),
        find_minimum(row + 1, col + 1, triangle)
    )
    return triangle[row][col] + minimum

triangle = [
    [1],
    [2,3],
    [4, 5, 6],
    [7,10,9,8],
    [11,12,13,14,15],
    list(range(16,22))]
minimum = find_minimum(0, 0, triangle)
print("The minimum cost is", minimum)
```

```python
def coin_change(coins, amount):
    changes = []
    largest = 0
    while amount > 0:
        if amount < coins[largest]:
            largest += 1
        else:
            changes.append(coins[largest])
            amount -= coins[largest]
    return changes
coins=[500,200,100,50,20,10,5,2,1]
amount=int(input('enter the amount'))
changes=coin_change(coins,amount)
print(changes,len(changes))
```
